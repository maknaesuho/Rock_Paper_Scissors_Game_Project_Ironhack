```python
import random
import time
choices = {1: "rock🪨", 2: "paper📃", 3: "scissors✂️"}

def welcome_message():
    """
    The function prints the welcome message and asks the player if they want to start the game
    
    """
    print("""
=====================================================
        ROCK - PAPER - SCISSORS CHAMPIONSHIP         
=====================================================
  Win 3 rounds against the computer to be champion!
=====================================================
""")
    time.sleep(2)
    start = input("Are you ready to start? Press Enter to begin the battle...")
    return start
```

```python
def get_player_input():
    """
    The function prompts the player to choose rock, paper, or scissors and validates the choice
    """
    player_input = input("Enter a choice rock🪨 - 1, paper📃 - 2, scissors✂️ - 3: ")
    while player_input not in ["1", "2", "3"]:
        print(" Please only enter a valid number: 1, 2 or 3")
        player_input = input("Enter a choice: rock🪨 - 1, paper📃 - 2, scissors✂️ - 3: ")
    return int(player_input)
```

```python
def get_computer_input():
    """
    The function randomises a number that would be computer's choice
    """
    computer_input = random.choice([1,2,3])
    return computer_input
```

```python
def process_turn(game_state):
    """
    The function executes the core game loop, collecting player moves, generating
    computer moves, evaluating round outcomes, and printing live round 
    results until either reaches 3 points
    
    Parameters:
    rounds: The current round number
    player_score: The player's score at the start of the turn sequence
    computer_score: The computer's score at the start of the turn sequence

    """
    
    while game_state["player_score"] < 3 and game_state["computer_score"] < 3:
        player_input = get_player_input()
        computer_input = get_computer_input()
        game_state["rounds"] += 1
        
        if player_input == computer_input:
            print(f"\n--- Round {game_state["rounds"]} ---")
            print(f"You both chose {choices[player_input]}") 
            print(f"Your score: {game_state["player_score"]} Computer score: {game_state["computer_score"]}\n")
        elif player_input == 1 and computer_input == 2:
            game_state["computer_score"] += 1
            print(f"\n--- Round {game_state["rounds"]} ---")
            print("Paper covers rock! You lose!") 
            print(f"Your score: {game_state["player_score"]} Computer score: {game_state["computer_score"]}\n")
        elif player_input == 2 and computer_input == 1:
            game_state["player_score"] += 1
            print(f"\n--- Round {game_state["rounds"]} ---")
            print("Paper covers rock! You win!")
            print(f"Your score: {game_state["player_score"]} Computer score: {game_state["computer_score"]}\n")    
        elif player_input == 3 and computer_input == 2:
            game_state["player_score"] += 1
            print(f"\n--- Round {game_state["rounds"]} ---")
            print("Scissors cuts paper! You win!")
            print(f"Your score: {game_state["player_score"]} Computer score: {game_state["computer_score"]}\n")    
        elif player_input == 1 and computer_input == 3:
            game_state["player_score"] += 1
            print(f"\n--- Round {game_state["rounds"]} ---")
            print("Rock smashes scissors! You win!")
            print(f"Your score: {game_state["player_score"]} Computer score: {game_state["computer_score"]}\n")
        elif player_input == 2 and computer_input == 3:
            game_state["computer_score"] += 1
            print(f"\n--- Round {game_state["rounds"]} ---")
            print("Scissors cuts paper! You lose!")
            print(f"Your score: {game_state["player_score"]} Computer score: {game_state["computer_score"]}\n")
        elif player_input == 3 and computer_input == 1:
            game_state["computer_score"] += 1
            print(f"\n--- Round {game_state["rounds"]} ---")
            print("Rock smashes scissors! You lose!")
            print(f"Your score: {game_state["player_score"]} Computer score: {game_state["computer_score"]}\n")

    return game_state
```

```python
def winner(game_state):
    """
    The function evaluates the final scores, announces the overall game winner 
    and prompts the player with validated input to replay or exit

    Parameters:
    player_score: The final score accumulated by the player
    computer_score: The final score accumulated by the computer
    
    """
    if game_state["player_score"] == 3:
        print("""
*****************************************************
               CONGRATULATIONS, YOU WON!
               🏆🏆🏆🏆🏆🏆🏆🏆🏆🏆
*****************************************************
""")
    elif game_state["computer_score"] == 3:
        print("""
-----------------------------------------------------
               GAME OVER - YOU LOST!
               🪦🪦🪦🪦🪦🪦🪦🪦
-----------------------------------------------------
""")
    play_again = input("Want to try again?(yes/no): \n").strip().lower()
    while play_again not in ["yes", "no"]:
        print("Please enter either 'yes' or 'no'.")
        play_again = input("Want to try again?(yes/no): \n").strip().lower()

    if play_again == "yes":
        start_game()
    else:
        print("Thanks for playing!")
        
    if play_again == "yes":
        start_game()
```

```python
def start_game():
    """
    The function initialises and manages the full gameplay loop by displaying the 
    welcome message, running rounds until a score limit is reached, 
    and determining the match outcome
    
    """
    
    game_state = {
        "rounds": 0,
        "player_score": 0,
        "computer_score": 0,
        }
    welcome_message()
    game_state = process_turn(game_state)
    winner(game_state)
```

```python
start_game()
```
